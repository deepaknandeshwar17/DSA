```python
# alternative elements of an array

def alter(arr):

    result = []

    for i in range(0,len(arr), 2):
        result.append(arr[i])
    return result

arr = [1,2,3,4,5,6,7,8,9]

alter(arr)
```

```python
# Leaders in an array
# Given an array arr[] of size n, the task is to find all the Leaders in the array. An element is a Leader if it is greater than or equal to all the elements to its right side.


def leaders(arr):

    result = []
    n = len(arr)

    for i in range(n):
        for j in range(i+1, len(arr)):
            if arr[i] < arr[j]:
                break
        else:
            result.append(arr[i])
    return result

arr = [16, 17, 4, 3, 5, 2]

leaders(arr)
        
```

```python
# Reverse an array

def revArray(arr):

    left = 0
    right = len(arr)-1

    while left < right :
        arr[left], arr[right] = arr[right], arr[left]

        left = left + 1
        right = right -1

    return arr

arr = [10,2,3,4,5]
# revArray(arr)


arr1 = arr[::-1]
# arr1

arr.reverse()
print(arr)
```

```python
def drop_duplicates(arr):
    seen = set()
    result = []

    for num in arr:
        if num not in seen:
            seen.add(num)
            result.append(num)

    return result
```
