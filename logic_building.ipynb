```python
# Even or odd number checker

def is_even(n):
    rem = n % 2;
    if rem == 0:
        return "Even"
    else:
        return "Odd"

if __name__ == "__main__":
    number = int(input("Enter a number: "))
    result = is_even(number)
    print(f"The number {number} is {result}.")
```

```python
# sum of natural numbers

def sum_of_n_nums(n):
    sum = 0
    i = 1
    while i <= n:
        sum = sum+i
        i=i+1
    return sum

# print(sum_of_n_nums(5))


def find_sum_of_n(n):
    if n == 0:
        return 0
    if n == 1:
        return 1
    return n + find_sum_of_n(n-1)

# find_sum_of_n(0)


def findSum(n):
    return n*(n+1)//2

findSum(3)
```

```python
# sum of n squares

def squareSum(n):
    return sum(i**2 for i in range(1,n+1))

# squareSum(8)

def findSumSquare(n):
    if n == 0:
        return 0
    if n == 1:
        return 1
    return n**2 + findSumSquare(n-1)

# findSumSquare(2)

def sumSquare(n):
    sum = 0
    i = 1

    while i <= n:
        sum = sum + i**2
        i = i+1
    
    return sum

sumSquare(10)
```

```python
# Swap numbers

def swapNum(a,b):
    temp = a
    a = b 
    b = temp
    # return a,b
    print(a,b)

# swapNum(10,20)


def swapN(a,b):
    a = a+b
    b = a-b
    a = a-b
    return (a,b)

# swapN(10,20)

def swapNu(a,b):
    a,b = b,a
    return (a,b)

swapNu(2,3)
```
